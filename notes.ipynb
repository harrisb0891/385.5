# Datetime modules
-   `datetime` - built-in module for datetime functionality
-   `calendar` - built-in module, calendar related functionality
-   `time` - module for time functionality

```python
from datetime import date, datetime, time
```

```python
# get today
today = date.today()

print(today)
print(today.year)
print(today.month)
print(today.day)
```

```python
# Create date object
last_year = date(2025, 9, 18)
print(last_year)
```

```python
today_to_the_sec = datetime.now()
print(today_to_the_sec)
print(today_to_the_sec.year)
print(today_to_the_sec.month)
print(today_to_the_sec.day)
print(today_to_the_sec.hour)
print(today_to_the_sec.minute)
print(today_to_the_sec.second)
print(today_to_the_sec.microsecond)
print(today_to_the_sec.tzinfo)

print(today_to_the_sec.timestamp())
```

```python
time_ish = time(hour=12, minute=45, second=45)
print(time_ish)
```
